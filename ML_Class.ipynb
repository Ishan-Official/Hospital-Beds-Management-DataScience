## Machine Learning

- Definition

Machine Learning (ML) is a branch of Artificial Intelligence (AI) that enables computers to learn patterns from data and make predictions or decisions without being explicitly programmed for every task.

- Types of Machine Learning

There are mainly 3 types:

- 1. Supervised Learning

In supervised learning, the model learns from labeled data.

Input Data → Model → Known Output

Example:

Hours Studied	Marks
2	40
4	60
6	75
8	90

The model learns to predict marks based on study hours.

Types:

Regression → predicts continuous values
Classification → predicts categories/classes

Algorithms:

Linear Regression
Logistic Regression
Decision Tree
Random Forest
SVM
K-Nearest Neighbors

- 2. Unsupervised Learning

In unsupervised learning, the data has no labeled output. The model tries to discover patterns or groups in the data.

Example:

Customer Data
      ↓
Machine Learning
      ↓
Customer Groups

Common algorithms:

K-Means Clustering
Hierarchical Clustering
DBSCAN
PCA

Applications:

Customer segmentation
Pattern discovery
Anomaly detection
Data visualization

 - 3. Reinforcement Learning

In reinforcement learning, an agent learns by interacting with an environment.

Agent → Action → Environment
  ↑                  ↓
  └──── Reward ──────┘

The agent receives:

Reward → good action
Penalty → bad action

Examples:

Game-playing AI
Robotics
Self-driving systems
Recommendation strategies

- Types of Data in Machine Learning

Data can be classified in several ways.

   - 1. Numerical Data

Data represented using numbers.

a. Discrete Data

Countable values.

Examples:

Number of students = 30
Number of cars = 5
Number of children = 2
b. Continuous Data

Can have decimal values.

Examples:

Height = 5.7 feet
Weight = 65.5 kg
Temperature = 27.6°C

   -  2. Categorical Data

Data representing categories or groups.

a. Nominal Data

Categories with no natural order.

Examples:

Gender: Male, Female
Color: Red, Blue, Green
City: Kathmandu, Pokhara, Biratnagar
b. Ordinal Data

Categories that have a meaningful order.

Examples:

Education:
School < Bachelor < Master < PhD
Rating:
Poor < Average < Good < Excellent

   - 3. Binary Data

Data with only two possible values.

Examples:

Yes / No
True / False
0 / 1
Pass / Fail

    - 4. Text Data

Data consisting of words or sentences.

Examples:

"I am learning Machine Learning."
"This product is very good."

Used in:

NLP
Sentiment analysis
Chatbots
Text classification

   - 5. Image Data

Data represented as images.

Examples:

Cat image
Face image
X-ray image
Handwritten digit

Used in:

Computer Vision
Face recognition
Object detection
Medical image analysis

   - 6. Time-Series Data

Data collected over time.

Example:

Date	Temperature
Jan 1	20°C
Jan 2	22°C
Jan 3	21°C

Examples:

Stock prices
Weather data
Sales data
Sensor data
