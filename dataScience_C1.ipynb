#### Population(N)

 In Data Science and Statistics, population means the complete group of people, objects, events, 
or data items that we are interested in studying.

Simple definition:

Population is the entire set of data or individuals about which we want to draw conclusions.

#### Sample(n):

In Data Science and Statistics, a sample is a smaller subset selected from a population for analysis.

Definition:

A sample is a part of the population that is selected to represent the entire population.

Here, n represents the sample size, meaning the number of observations/items in the sample.

Example

Suppose a university has 10,000 students:

Population (N) = 10,000 students
Sample (n) = 500 selected students
N = Population size
n = Sample size

##### Inferential Statistics

2. Inferential Statistics

Definition:

Inferential statistics is used to draw conclusions or make predictions about a population using sample data.

It helps us answer questions about a large population without studying everyone.

Examples:

Hypothesis testing
Confidence intervals

Prediction

### Measures of Central Tendency

Measures of Central Tendency

Measures of Central Tendency are statistical methods used to find the central or typical value of a dataset.

There are 3 main measures:

1. Mean

Mean is the average of all values.

Formula:

$$ Mean = \frac{\sum X}{N} $$

Example:

Data: 10, 20, 30, 40, 50

$$ Mean = \frac{10+20+30+40+50}{5}=30 $$

👉 Mean = 30

2. Median

First, arrange the data in ascending order.

If \(n\) is odd:
$$ \boxed{Median = \left(\frac{n+1}{2}\right)^{th}\text{ value}} $$
If \(n\) is even:
$$ \boxed{Median=\frac{\left(\frac n2\right)^{th}\text{ value}+\left(\frac n2+1\right)^{th}\text{ value}}{2}} $$

Example: 10, 20, 30, 40, 50

3. Mode

There is no general mathematical formula for simple ungrouped data.

$$ \boxed{Mode=\text{Value with the highest frequency}} $$

Example:
10, 20, 20, 30, 40

$$ \boxed{Mode=20} $$

because 20 occurs twice, while the other values occur once.

Quick Comparison
Measure	Meaning	Example
Mean	Average value	30
Median	Middle value	30

Easy way to remember

Mean → Average
Median → Middle
Mode → Most repeated

These three are collectively called measures of central tendency.

IQR stands for Interquartile Range. It measures the spread of the middle 50% of data.

Formula

### Measure of shape(Skewness)

Skewness is a statistical measure that tells us about the shape and asymmetry of a data distribution.

In simple words:

Skewness tells us whether the data is distributed symmetrically or has a longer tail on one side.

Measure of Shape — Skewness Formula
1. Population Skewness

The formula for population skewness is:

$$ \boxed{ \gamma_1 = \frac{\frac{1}{N}\sum_{i=1}^{N}(x_i-\mu)^3} {\sigma^3} } $$

Where:

\(x_i\) = each data value
\(\mu\) = population mean
\(N\) = number of observations
\(\sigma\) = population standard deviation
\(\gamma_1\) = skewness
2. Sample Skewness

A commonly used sample formula is:

$$ \boxed{ G_1 = \frac{n}{(n-1)(n-2)} \sum \left(\frac{x_i-\bar{x}}{s}\right)^3 } $$

Where:

\(n\) = sample size
\(\bar{x}\) = sample mean
\(s\) = sample standard deviation
Simple interpretation
$$ \boxed{\text{Skewness}=\frac{\text{Third central moment}}{(\text{Standard deviation})^3}} $$
Skewness	Shape
\(=0\)	Symmetric
\(>0\)	Positively/right skewed
\(<0\)	Negatively/left skewed
Python

Types of Skewness
1. Symmetric Distribution

When data is approximately equally distributed on both sides.

        /\
       /  \
      /    \
_____/      \_____

Skewness ≈ 0

Example:

data = [20, 25, 30, 35, 40]
2. Positive Skewness (Right-Skewed)

The distribution has a longer tail toward the right.

       /\
      /  \
     /    \____
____/          \________

Skewness > 0

Usually:

Mean > Median > Mode

Example: Income distribution — a small number of people may have very high incomes.

3. Negative Skewness (Left-Skewed)

The distribution has a longer tail toward the left.

________
       /\
      /  \
_____/    \

Skewness < 0

Usually:

Mean < Median < Mode

Example: An exam where most students score very high but a few students score very low.

Remember
Skewness = Measure of asymmetry

Positive → Tail on RIGHT →
Negative → Tail on LEFT  ←
Zero     → Approximately symmetric
