#### Population(N)

 In Data Science and Statistics, population means the complete group of people, objects, events, 
or data items that we are interested in studying.

Simple definition:

Population is the entire set of data or individuals about which we want to draw conclusions.

#### Sample(n):

In Data Science and Statistics, a sample is a smaller subset selected from a population for analysis.

Definition:

A sample is a part of the population that is selected to represent the entire population.

Here, n represents the sample size, meaning the number of observations/items in the sample.

Example

Suppose a university has 10,000 students:

Population (N) = 10,000 students
Sample (n) = 500 selected students
N = Population size
n = Sample size


## Statistics

Statistics in Data Science

Definition:

Statistics is the science of collecting, organizing, analyzing, interpreting, and presenting data to make meaningful conclusions and decisions.

Example

Suppose you have the exam marks of 100 students:

45, 67, 72, 89, 56, 78, ...



Types of Statistics

1. Descriptive Statistics
2. Inferential Statistics


In short:

### Descriptive statistic

1. Descriptive Statistics

Definition:

Descriptive statistics is used to collect, organize, summarize, and present data.

It tells us what the data looks like.

types:
- Measures of central tendency
- Measures of variability
- Percentiles
- Measures of shape
- Frequency Disribution
- Covariance and Correlation

Examples:

Mean
Median
Mode
Range
Variance
Standard deviation
Tables
Graphs and charts

Example:
Marks of 5 students:

50, 60, 70, 80, 90

Mean:

(50 + 60 + 70 + 80 + 90) / 5 = 70

So, 70 is the descriptive statistic (mean) of the data.



### Inferential Statistics

2. Inferential Statistics

Definition:

Inferential statistics is used to draw conclusions or make predictions about a population using sample data.

It helps us answer questions about a large population without studying everyone.

Examples:

Hypothesis testing
Confidence intervals
Regression analysis
Correlation
ANOVA
Prediction

Example:

A university has 10,000 students. We select 500 students and calculate their average marks.

Population = 10,000 students
Sample = 500 students

We use the sample results to make conclusions about the 10,000 students.

Simple Difference
Descriptive Statistics	Inferential Statistics
Describes the data	Makes conclusions about a population
Uses data we have	Uses sample data to generalize
Mean, median, mode	Hypothesis testing, confidence interval
"What happened?"	"What can we conclude?"

Remember:
👉 Descriptive = Describe the data
👉 Inferential = Infer/conclude about the population

#### Measures of Central Tendency

Measures of Central Tendency

Measures of Central Tendency are statistical methods used to find the central or typical value of a dataset.

There are 3 main measures:

1. Mean

Mean is the average of all values.

Formula:

$$ Mean = \frac{\sum X}{N} $$

Example:

Data: 10, 20, 30, 40, 50

$$ Mean = \frac{10+20+30+40+50}{5}=30 $$

👉 Mean = 30

2. Median

Median is the middle value when data is arranged in ascending or descending order.

Example:

10, 20, 30, 40, 50

👉 Median = 30

For an even number of values:

10, 20, 30, 40

$$ Median = \frac{20+30}{2}=25 $$

👉 Median = 25

3. Mode

Mode is the value that occurs most frequently in a dataset.

Example:

10, 20, 20, 30, 40

👉 Mode = 20

because 20 occurs twice, while the other values occur once.

Quick Comparison
Measure	Meaning	Example
Mean	Average value	30
Median	Middle value	30
Mode	Most frequent value	20
Easy way to remember

Mean → Average
Median → Middle
Mode → Most repeated

These three are collectively called measures of central tendency.
